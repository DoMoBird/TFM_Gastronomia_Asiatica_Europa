## 1. Importar librerías

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 150)
```

## 2. Cargar el dataset completo

```python
DATA_PATH = '../data/tripadvisor_european_restaurants.csv'

df = pd.read_csv(DATA_PATH, low_memory=False)
print(f'Shape original: {df.shape}')
df.head()
```

## 3. Filtrar restaurantes asiáticos

La columna `cuisines` contiene una lista de tipos de cocina separados por comas (ej. `"Chinese, Asian, Sushi"`).
Filtramos las filas donde alguna de las cocinas asiáticas definidas aparezca en esa columna.

```python
asian_cuisines = ['Chinese', 'Japanese', 'Thai', 'Korean',
                  'Vietnamese', 'Indian', 'Asian', 'Sushi',
                  'Dim Sum', 'Ramen', 'Noodles']

pattern = '|'.join(asian_cuisines)
mask = df['cuisines'].str.contains(pattern, case=False, na=False, regex=True)

df_asian = df[mask].copy()
print(f'Restaurantes asiáticos: {df_asian.shape[0]} de {df.shape[0]} totales ({df_asian.shape[0] / df.shape[0]:.2%})')
```

## 4. Primera exploración del subconjunto filtrado

```python
# Shape y columnas
print('Shape:', df_asian.shape)
print('\nColumnas:')
print(df_asian.columns.tolist())
```

```python
# Nulos por columna (%)
nulos = (df_asian.isna().sum() / len(df_asian) * 100).sort_values(ascending=False)
nulos[nulos > 0]
```
