## 1. Introducción y motivación

La gastronomía asiática (china, japonesa, tailandesa, coreana, vietnamita, india, entre otras) es una
de las categorías con mayor presencia y crecimiento en el sector de la restauración europea de las
últimas décadas. Sin embargo, la información sobre su distribución geográfica, la diversidad de
propuestas culinarias y los factores que determinan el éxito de un restaurante suele estar dispersa
y no analizada de forma sistemática.

Este TFM tiene como objetivo caracterizar la presencia de la gastronomía asiática en Europa, con
especial atención a España, y **construir un modelo predictivo capaz de estimar la valoración media
que recibiría un restaurante asiático** en función de sus características (tipo de cocina, ubicación,
rango de precio, sub-puntuaciones de comida/servicio/ambiente, etc.).

**Objetivos específicos:**

1. Analizar la distribución geográfica y los tipos de cocina asiática predominantes en Europa.
2. Identificar qué variables influyen más en la valoración (`rating`) de un restaurante.
3. Construir y comparar varios modelos de Machine Learning para predecir esa valoración.
4. Interpretar el modelo ganador con técnicas de explicabilidad (SHAP).
5. Productivizar el modelo en una función reutilizable que estime la valoración de un restaurante
   nuevo a partir de sus características.

El enfoque combina técnicas vistas a lo largo del máster: estadística descriptiva y regresión
(Minería de Datos), modelos de árboles/ensemble y SVM (Machine Learning) y redes neuronales
(Deep Learning), aplicadas de extremo a extremo sobre un caso de negocio real.

## 4. Preprocesamiento y transformación de datos

Los datos brutos, tal como se descargan, no están listos para entrenar un modelo: hay huecos sin
rellenar, columnas que no aportan información útil (o que directamente "hacen trampa", como se ve
más abajo) y variables de texto que hay que convertir a números. Este apartado resume ese trabajo de
limpieza y transformación.

**Limpieza:**
- Se eliminan las 4.991 filas sin `avg_rating` (variable objetivo), quedando 81.315 restaurantes.
- Se descartan columnas de texto libre no estructurado (`keywords`, `top_tags`, `awards`, `features`,
  dirección, enlaces...), columnas con fuga de información (`excellent`, `very_good`, `average`,
  `poor`, `terrible` — son conteos derivados del propio rating, así que dárselas al modelo sería
  "regalarle la respuesta") y columnas de altísima cardinalidad sin procesar (`city`, `province`).
- **Hallazgo relevante:** `popularity_generic` resultó ser, tras inspección, un identificador de
  texto casi único por restaurante (81.161 valores distintos en 81.315 filas, del tipo
  `"#5907 of 18480 places to eat in Paris"`). Incluirlo como variable categórica desestabilizaba
  por completo el modelo de Regresión Lineal (generaba un R² negativo, es decir, peor que no predecir
  nada); además, el ranking de TripAdvisor se calcula a partir del propio rating, por lo que también
  sería fuga de información. Se excluyó del modelo.
- La variable `total_reviews_count`, muy asimétrica (la mayoría de restaurantes tiene pocas reseñas y
  unos pocos tienen miles), se transforma matemáticamente (`log1p`) en vez de eliminar los casos
  extremos, para no perder información.

![Outliers de reseñas](data/fig_outliers_reviews.png)

**Feature engineering** (construcción de variables a partir de los datos originales, para que el
modelo pueda aprovecharlas):
- Variables binarias (Sí/No → 1/0): reclamado (`claimed`), apto vegetariano/vegano/sin gluten.
- `price_level` (€ / €€-€€€ / €€€€) codificado como variable ordinal (1-3), es decir, como un número
  que respeta el orden creciente de precio.
- `meals` (Lunch, Dinner, Breakfast...) y `cuisines` (tipos de cocina asiática) transformados en
  columnas Sí/No independientes (p. ej. "¿sirve cena?", "¿es cocina china?"), más un conteo
  `n_cuisines` del número de cocinas ofrecidas.
- País agrupado en los 15 más frecuentes + categoría "Otros", para evitar crear cientos de columnas
  minúsculas por países con muy pocos restaurantes.

**Pipeline final** (la secuencia de pasos que transforma los datos brutos en la tabla que ve el
modelo): relleno de huecos (con la mediana en variables numéricas, con 0 en binarias, con el valor
más frecuente en categóricas) + puesta en una escala común de las variables numéricas + conversión de
categorías de texto en columnas 0/1 (*one-hot encoding*). Todo el proceso se ajusta **únicamente**
con el 80 % de los datos reservado para entrenar, y se aplica igual al 20 % restante reservado para
evaluar (`random_state=42`) — así se evita que el modelo "haga trampa" viendo de antemano los datos
con los que luego será examinado.

## 5. Modelización

### 5.1 Predicción de valoración de restaurantes

Se plantea un problema de **regresión supervisada**: predecir `avg_rating` (escala 1-5) a partir de
las características del restaurante (ubicación, tipo de cocina, rango de precio, sub-puntuaciones,
horarios, etc.). El modelo se evalúa siempre sobre un **20 % de los datos que nunca ha visto durante
el entrenamiento** (el "examen"), para asegurar que el resultado refleja cómo se comportaría con
restaurantes nuevos y no simplemente que ha memorizado los datos de entrenamiento.

> **Cómo leer las métricas de esta sección:**
> - **RMSE** y **MAE**: el error típico de la predicción, expresado en la misma escala que el rating
>   (1 a 5). Un MAE de 0,36, por ejemplo, significa que en promedio el modelo se equivoca en ±0,36
>   puntos sobre 5 — cuanto más bajo, mejor.
> - **R²**: qué proporción de las diferencias de valoración entre restaurantes logra explicar el
>   modelo, en una escala de 0 (no explica nada) a 1 (explica todo perfectamente). Un R² de 0,40
>   significa que el modelo capta una parte real pero no toda la variabilidad — el resto depende de
>   factores no recogidos en los datos (por ejemplo, gustos subjetivos de cada cliente).

### 5.2 Comparativa de modelos

Se entrenan y comparan cinco modelos, eligiendo deliberadamente algoritmos vistos y practicados con
código durante las distintas asignaturas del máster (Minería de Datos, Machine Learning y Deep
Learning):

| Modelo | RMSE | MAE | R² | Asignatura |
|---|---|---|---|---|
| Regresión Lineal | 0,548 | 0,375 | 0,372 | Minería de Datos |
| Random Forest | 0,557 | 0,365 | 0,349 | Machine Learning |
| XGBoost (sin ajustar) | 0,548 | 0,363 | 0,371 | Machine Learning |
| SVR (Support Vector Regression) | 0,560 | 0,368 | 0,344 | Machine Learning |
| Red densa (Keras) | 0,540 | 0,362 | 0,389 | Deep Learning |

![Comparativa de modelos](data/fig_comparativa_modelos.png)

Los cinco modelos obtienen resultados similares (R² entre 0,34 y 0,39), lo que indica que la relación
entre las variables disponibles y la valoración es en gran parte lineal, dominada por las
sub-puntuaciones de comida, servicio y valor.

**Modelo elegido para productivizar: XGBoost.** Su rendimiento es prácticamente idéntico al de la
red neuronal (diferencia de RMSE < 0,015, es decir, menos de 2 centésimas sobre una escala de 5), pero
permite interpretabilidad nativa y rápida (ver 5.3) y no requiere el entorno de ejecución específico
que necesita TensorFlow.

Tras un **ajuste de hiperparámetros** (probar de forma sistemática distintas configuraciones internas
del modelo — 30 combinaciones, cada una validada con **validación cruzada**: se entrena y evalúa 5
veces sobre 5 particiones distintas de los datos, para confirmar que el resultado no depende de qué
filas cayeron por azar en el conjunto de test) el RMSE de XGBoost mejora de 0,548 a **0,536**, y el R²
sube de 0,371 a **0,399**. Esa misma validación cruzada confirma que el modelo es estable: el error
apenas varía entre las 5 particiones (RMSE = 0,534 ± 0,004).

![Estabilidad entre folds](data/fig_cv_stability.png)

### 5.3 Interpretabilidad del modelo (SHAP)

Un modelo que predice bien pero no explica *por qué* es difícil de defender ante negocio. **SHAP** es
una técnica que, para cada predicción, reparte el "mérito" entre las variables de entrada — permite
responder a la pregunta "¿qué ha hecho que el modelo prediga este rating y no otro?" tanto de forma
agregada (qué variables importan en general) como para un restaurante concreto.

![SHAP summary](data/fig_shap_summary.png)
![SHAP importancia](data/fig_shap_bar.png)

**Conclusión de la interpretabilidad:** las variables con mayor impacto en la valoración predicha
son, por este orden, `food`, `service` y `value`, seguidas de `atmosphere` y el volumen de reseñas.
El país y el tipo de cocina matizan la predicción pero con un peso claramente menor. Esto es coherente
con la intuición de negocio: la calidad percibida de la comida y el servicio son, con diferencia, los
factores que más determinan la reputación de un restaurante, por encima de dónde esté ubicado o qué
tipo de cocina asiática ofrezca.

## 6. Productivización del modelo

El pipeline completo (preprocesamiento + modelo XGBoost afinado) se guarda con `joblib` y se envuelve
en una función `predict_rating()` que recibe las características de un restaurante en lenguaje
natural (sin que el usuario tenga que conocer el feature engineering interno) y devuelve la
valoración media estimada:

```python
predict_rating(
    cuisines=['Japanese', 'Sushi'], country='Spain', price_level='€€€€',
    total_reviews_count=400, food=4.5, service=4.5, value=4.0, atmosphere=4.5,
)
# → 4.32 / 5
```

Ejemplos de uso y resultado:

| Escenario | Predicción |
|---|---|
| Japonés premium, buenas sub-puntuaciones, España | 4,32 / 5 |
| Chino económico, pocas reseñas, sub-puntuaciones mediocres, Reino Unido | 2,83 / 5 |
| Tailandés en Francia (sin especificar sub-puntuaciones, valores por defecto) | 4,00 / 5 |

Cualquier parámetro no especificado toma automáticamente el valor mediano/modal del dataset de
entrenamiento, por lo que la función es robusta ante información incompleta — un caso de uso
realista, ya que un emprendedor evaluando la apertura de un restaurante nuevo raramente tendrá todos
los datos de antemano.

Como línea de trabajo inmediata (fuera del alcance temporal de este TFM), esta función podría
exponerse como una pequeña aplicación web con Streamlit o una API con FastAPI, ambas herramientas
disponibles en el entorno de trabajo.

## 7. Conclusiones y líneas futuras

**Conclusiones principales:**

- La gastronomía asiática tiene una presencia significativa y heterogénea en Europa, concentrada en
  Reino Unido, Francia, Alemania, España e Italia, con Madrid como epicentro claro dentro de España.
- La cocina china y la etiqueta genérica "asiática" dominan la oferta, aunque con variaciones
  relevantes por país (p. ej. mayor peso de la cocina india en Reino Unido).
- La valoración de un restaurante asiático está determinada principalmente por la calidad percibida
  de la comida, el servicio y la relación calidad-precio — no por su ubicación geográfica ni por el
  tipo concreto de cocina asiática que ofrezca.
- Un modelo XGBoost, con hiperparámetros ajustados, predice la valoración media con un error medio
  (MAE) de 0,355 puntos sobre una escala de 1 a 5 — suficientemente preciso para un caso de uso de
  apoyo a la decisión (p. ej. estimar el potencial de éxito de un restaurante antes de abrirlo).

**Líneas futuras:**

1. **Actualidad de los datos:** este dataset fue recopilado en 2021 y no se ha vuelto a actualizar.
   Se buscó explícitamente una fuente pública más cercana a 2026 con escala y detalle comparables
   (más de 80.000 restaurantes asiáticos, con tipo de cocina, rating desglosado y geolocalización) y
   no se encontró ninguna — la alternativa más reciente localizada en Kaggle resultó tener solo 7
   restaurantes y 1.502 reseñas, insuficiente para sustituir o complementar de forma fiable el
   análisis. Como continuación de este trabajo, se proponen tres vías: (a) series de interés de
   búsqueda en Google Trends como proxy de demanda reciente; (b) conteos actuales de restaurantes vía
   Google Places API u OpenStreetMap, comparables como "foto fija" 2026 frente a la de este dataset
   (2020-2021); (c) un scraping propio con fecha de alta del restaurante, que permitiría reconstruir
   una serie temporal real de aperturas.
2. **Explotación del texto libre:** el dataset incluye una columna `keywords` con extractos de
   reseñas que no se ha explotado en este TFM. Aplicar técnicas de NLP (embeddings, fine-tuning de un
   modelo tipo DistilBERT, visto en la asignatura de Inteligencia Artificial) permitiría incorporar
   análisis de sentimiento como feature adicional del modelo.
3. **Despliegue real:** convertir la función `predict_rating()` en una aplicación web (Streamlit) o
   API (FastAPI) accesible públicamente, con la interpretabilidad SHAP integrada para explicar cada
   predicción al usuario final.
